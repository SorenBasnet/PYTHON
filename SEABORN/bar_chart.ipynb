```python
import pandas as pd 
import seaborn as sns 
import matplotlib.pyplot as plt 
```

```python
df = pd.read_csv("/Users/sorenbasnet/Documents/Github/PYTHON/STAT_443/data/alzheimers_disease_data.csv")
```

```python
eth_counts = df['Ethnicity'].value_counts(normalize=True)
eth_counts.plot(kind='bar')
plt.ylabel('Proportion')
plt.title('Ethnicity Proportions')
plt.show()
```

```python
# This can be now used to see how ethnicities relates to the outcome 

sns.countplot(x='Ethnicity', hue='Diagnosis', data=df)
plt.xticks()
plt.title('Ethnicity vs Outcome')
plt.show()
```

```python
eth_outcome = pd.crosstab(df['Ethnicity'], df['Ethnicity'])
eth_outcome.plot(kind='bar', stacked=True)
plt.title('Stacked Bar: Ethnicity vs Outcome')
plt.show()
```

```python
# Treemap (proportions with area):
import squarify

eth_counts = df['Ethnicity'].value_counts()
squarify.plot(sizes=eth_counts, label=eth_counts.index, alpha=0.8)
plt.axis('off')
plt.title('Ethnicity Treemap')
plt.show()
```

```python
import plotly.express as px

fig = px.bar(df['Ethnicity'].value_counts().reset_index(),
             x='index', y='ethnicity',
             labels={'index':'Ethnicity', 'ethnicity':'Count'},
             title='Ethnicity Counts')
fig.show()
```
